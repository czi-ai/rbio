## 1.1 TF: PMIs - Gene Pointwise Mutual Information Scores

**1. Binary confidence scores** <br>
**2. Soft confidence scores**

```python
tf_pmi_binary_df = pd.read_csv(f'{TF_DATA_DIR}TF_PMIs_sig_0.01-train-v0.0.3_binary.csv')
tf_pmi_soft_df = pd.read_csv(f'{TF_DATA_DIR}TF_PMIs_sig_0.01-train-v0.0.3_soft.csv')

get_confs_as_floats = lambda x: [eval(y) for y in x.split('|')]

tf_pmi_binary_df['class_confs_float'] = tf_pmi_binary_df['class_confidences'].apply(get_confs_as_floats)
tf_pmi_soft_df['class_confs_float'] = tf_pmi_soft_df['class_confidences'].apply(get_confs_as_floats)
```

```python
tf_pmi_binary_df.head(2)
```

```python
tf_pmi_soft_df.head(2)
```

```python
tf_pmi_binary_df.groupby('label').count()
```

```python
tf_pmi_soft_df.groupby('label').count()
```

```python
def get_class_confidences_interval(df):
    all_class_confidences = set([])
    class_confidences = df['class_confs_float'].to_list()
    for cc in class_confidences:
        valid_confs = set([x for x in cc if x!=0])
        all_class_confidences.update(valid_confs)
    min_conf = min(all_class_confidences)
    max_conf = max(all_class_confidences)
    print(f'Min class confidence: {min_conf} | Max class confidence: {max_conf}')
```

```python
get_class_confidences_interval(tf_pmi_binary_df)
```

```python
get_class_confidences_interval(tf_pmi_soft_df)
```
